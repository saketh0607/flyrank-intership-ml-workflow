## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import numpy as np
import pandas as pd
import os
if not os.path.exists("data"):
    os.makedirs("data")
if not os.path.exists("data/customer_churn_starter.csv"):
    synthetic_data = ({
        "account_age_months": np.random.randint(1, 48, size=500),
        "monthly_usage_hours": np.random.randint(100, 1000, size=500),
        "support_tickets_count": np.random.randint(0, 100, size=500),
        "is_churned": np.random.choice([0, 1], size=500, p=[0.8,0.2])
    })
    synthetic_data_df = pd.DataFrame(synthetic_data)
    synthetic_data_df.to_csv("data/customer_churn_starter.csv", index=False)

# --- Humanized Data Loading & Extraction Pipeline ---

target_data_path = "data/customer_churn_starter.csv"
loaded_customer_df = pd.read_csv(target_data_path)

# Extract core, high-level metrics from the data frame matrix
total_recorded_observations = len(loaded_customer_df)
total_missing_values_count = loaded_customer_df.isnull().sum().sum()

# Determine the exact ratio of churned vs retained entries
churn_column_series = loaded_customer_df["is_churned"]
total_churned_cases = int(churn_column_series.sum())
overall_churn_rate = (total_churned_cases / total_recorded_observations) * 100

# Present our numbers with explicit, easily scannable descriptions
print("---Exploratory Data Analysis Sanity Check ---")
print(f"Total Rows Processed: {total_recorded_observations}")
print(f"Total Missing Values: {total_missing_values_count}")
print(f"Active Historical Churn Count: {total_churned_cases} accounts")
print(f"Baseline System Churn Rate: {overall_churn_rate:.2f}%")
```
