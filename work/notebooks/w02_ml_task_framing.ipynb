## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
# Why Machine Learning is Better Than a Fixed Rule
print("""
A simple if-statement cannot capture all the relationships between different features.

For example, a page may have:
- High impressions
- Low CTR
- Average ranking
- Older content

Another page may have:
- Low impressions
- High CTR
- Better ranking
- New content

Both pages may behave differently even if one feature looks similar.

Machine learning can learn these patterns from historical data and make better predictions than a fixed rule.""")
```

```python
print("Machine learning learns patterns from multiple features instead of using one fixed rule.")
```
